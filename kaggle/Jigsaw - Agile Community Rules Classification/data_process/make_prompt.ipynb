```python
import pandas as pd
import sys
import os
import numpy as np
import json
```

```python
data = pd.read_csv("../data/train.csv")
```

```python
SYSTEM_prompt = """
### Task Description  
Your task is to determine whether a comment violates a specific rule of a subreddit. The training dataset contains only 2 rules, while the test dataset may include additional unseen rules (you must generalize your judgment based on the rule description and examples).  

### Input Structure  
- ​**Subreddit**: The name of the subreddit where the target comment was posted .  
- ​**Rule**: The specific rule to check for violation (the current subreddit's rule).  
- ​**Examples**​ (to assist in determining rule boundaries):  
  - Positive examples (Violation: Yes): Comments that clearly violate the rule (2 examples total).  
  - Negative examples (Violation: No): Comments that clearly do not violate the rule (2 examples total).  
- ​**Comment to Evaluate**: The content of the comment that needs to be judged for rule violation (body).  
"""
```

```python
make_data = []
for i, row in data.iterrows():
    text = f"""
r/{row.subreddit}
Rule: {row.rule}

1) {row.positive_example_1}
Violation: Yes

2) {row.negative_example_1}
Violation: No

3) {row.negative_example_2}
Violation: No

4) {row.positive_example_2}
Violation: Yes

5) {row.body}
"""
    make_data.append(text)
```

```python
output_path = "../data/fine_tune_data.jsonl"
with open(output_path, "w", encoding="utf-8") as f:
    for item in make_data:
        json.dump({"prompt": item}, f, ensure_ascii=False)
        f.write("\n")

print(f"微调数据已保存到 {output_path}")
```
